```python
# Resume Setup and PDF Extraction
This notebook extracts the resume text from the PDF and documents setup steps for the portfolio project.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">
from pathlib import Path
from PyPDF2 import PdfReader
path = Path('Resume-LikedIn-Ajit.pdf')
reader = PdfReader(path)
text = []
for i, page in enumerate(reader.pages):
    text.append(f"---PAGE {i+1}")
    text.append(page.extract_text() or '')
print('\n'.join(text))
```
