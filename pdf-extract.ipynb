# PDF Resume Text Extraction
Use Python to read the resume PDF and extract text.

```python
from pathlib import Path
workspace = Path(r'C:\Users\ajitk\web-portfolio')
resume_src = workspace / 'Resume-LikedIn-Ajit.pdf'
public_dest = workspace / 'public' / 'Resume-LikedIn-Ajit.pdf'
public_dest.parent.mkdir(parents=True, exist_ok=True)
with resume_src.open('rb') as fsrc, public_dest.open('wb') as fdst:
    fdst.write(fsrc.read())
print('resume copied to public', public_dest.exists())
```
